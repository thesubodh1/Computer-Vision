```python
# face blurring and saving images
import cv2
import mediapipe as mp

face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades+"haarcascade_frontalface_default.xml")

img = cv2.imread("./Data/dummy.jpg")
grey_img = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
face = face_cascade.detectMultiScale(img,scaleFactor=1.1,minNeighbors=5)

for (x,y,w,h) in face:
    img[y:y+h,x:x+w,:] = cv2.blur(img[y:y+h,x:x+w,:],(30,10))
cv2.imwrite("./Data/blurred_face.jpg",img)
cv2.imshow("frame",img)
cv2.waitKey(0)
```
